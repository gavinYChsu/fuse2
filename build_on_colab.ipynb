This notebook allows you to build the Linux executable for Fuse using Google Colab's free resources.

### Step 1: Clone the Repository
Replace the URL below with your fork if you have executed changes.

### Step 2: Select Build Variant and Install Dependencies
Choose whether you want the standard CPU version or the CUDA version (requires NVIDIA GPU to run).

```python
#@title Build Configuration
build_variant = "cuda" #@param ["cpu", "cuda"]

!sudo apt-get update
!sudo apt-get install -y libgl1-mesa-glx libglib2.0-0

requirements_file = "requirements.txt" if build_variant == "cpu" else "requirements-cuda.txt"
print(f"Installing dependencies from {requirements_file}...")

!pip install -r {requirements_file}
!pip install pyinstaller
```

### Step 3: Build Executable
Run PyInstaller to create the standalone Linux binary.

```python
executable_name = f"fuse_linux_{build_variant}"
!pyinstaller --name {executable_name} --onefile --clean --add-data "fuse:fuse" run.py
```

### Step 4: Download
The executable is now in the `dist` folder. Run this cell to download it.

```python
from google.colab import files
files.download(f'dist/{executable_name}')
```
